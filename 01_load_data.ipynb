## Python 入门②：交叉表

```python
pd.crosstab(sample['P_METHOD'],sample['A6'], normalize='index').round(3)#crossstab()交叉表功能,
                                                                        #normalize='index'--parameter参数，附加功能
                                                                        #normalize--按比例格式，index--行，每个数/行总和
                                                                        #round()3--保留3位小数
```

```python
tab=pd.crosstab(sample['P_METHOD'],sample['A6'], normalize='index')#存进tab
tab #显示给我看 意思
```

```python
(tab[4.0]+tab[5.0]).round(3)#取列series，pandas会按index行配对相加，4.0--数字不用''
```

```python
(tab[1.0]+tab[2.0]).round(3)
```
